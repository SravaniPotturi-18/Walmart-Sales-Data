```python
df = df.drop_duplicates()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

```python
df["date"].dtype
```

```python
numeric_columns = [
    "store",
    "temperature",
    "fuel_price",
    "markdown1",
    "markdown2",
    "markdown3",
    "markdown4",
    "markdown5",
    "cpi",
    "unemployment"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

```python
df.columns.tolist()
```

```python
df["isholiday"].value_counts()
```

```python
df["isholiday"] = (
    df["isholiday"]
    .astype(str)
    .str.strip()
    .str.upper()
)

df["isholiday"].value_counts()
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
markdown_columns = [
    "markdown1",
    "markdown2",
    "markdown3",
    "markdown4",
    "markdown5"
]

df[markdown_columns] = df[markdown_columns].fillna(0)
```

```python
df[markdown_columns].isnull().sum()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["cpi"] = df["cpi"].fillna(df["cpi"].median())

df["unemployment"] = df["unemployment"].fillna(
    df["unemployment"].median()
)
```

```python
df[["cpi", "unemployment"]].isnull().sum()
```

```python
df.duplicated().sum()
```

```python
output_file = "cleaned_dataset.csv"

df.to_csv(
    output_file,
    index=False
)

print("Cleaned dataset saved successfully!")
```
